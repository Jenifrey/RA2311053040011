```python
import pandas as pd

data = pd.read_csv('/kaggle/input/datasets/nishantttt/stock-dataset/NSEI (3).csv')

print(data.head())
```

```python
data['Date'] = pd.to_datetime(data['Date'])
data = data.sort_values('Date')
data = data.dropna()

print("Data cleaned successfully")
```

```python
print(data.describe())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.plot(data['Date'], data['Close'])
plt.title("Stock Trend Analysis (NSEI)")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
data['MA50'] = data['Close'].rolling(50).mean()

plt.figure(figsize=(12,6))
plt.plot(data['Date'], data['Close'], label='Close Price')
plt.plot(data['Date'], data['MA50'], label='50-day Moving Average')
plt.title("Stock Trend with Moving Average")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.grid()
plt.show()
```

The stock shows an overall increasing trend over time.
Moving average smooths fluctuations and highlights long-term trends.
Time series analysis helps understand stock behavior effectively.
